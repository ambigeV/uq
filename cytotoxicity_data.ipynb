```python
import pandas as pd
import pickle
import os

# --- SETUP ---
DATA_FOLDER = 'cytotoxicity_data'

file_pairs = [
    ('HEK293_train_BM.pkl', 'HEK293_train_BM_AC.pkl'),
    ('HEK293_test_BM.pkl',  'HEK293_test_BM_AC.pkl'),
    ('tox21_all.pkl',       'tox21_all_AC.pkl')
]

def load_pkl(file_name):
    with open(os.path.join(DATA_FOLDER, file_name), 'rb') as f:
        # Using latin1 to handle NumPy/Pandas cross-version compatibility
        return pickle.load(f, encoding='latin1')

# --- PROCESSING ---
for orig_name, ac_name in file_pairs:
    print(f"📦 Processing: {orig_name}")
    
    df_orig = load_pkl(orig_name)
    df_ac = load_pkl(ac_name)
    
    # Identify unique columns in the AC version
    new_cols = [c for c in df_ac.columns if c not in df_orig.columns]
    
    # 1. SPECIAL CASE: tox21 (Positional Alignment)
    if 'tox21' in orig_name:
        print("   🔗 Applying Positional Alignment (Row-by-Row)")
        # Since rows align perfectly, we can just join on the index
        # We take only the new columns from df_ac
        df_aca = pd.concat([df_orig, df_ac[new_cols]], axis=1)
        
    # 2. STANDARD CASE: HEK (CID Alignment)
    else:
        print("   🔗 Applying CID-based Alignment")
        # Find the CID column name (handling potential case differences)
        cid_col = next((c for c in df_orig.columns if 'pubchem' in str(c).lower()), 'PUBCHEM_CID')
        
        # Merge ensuring original order is preserved
        df_aca = pd.merge(
            df_orig, 
            df_ac[[cid_col] + new_cols], 
            on=cid_col, 
            how='left'
        )

    # --- SAVE RESULT ---
    aca_name = ac_name.replace('_AC.pkl', '_ACA.pkl')
    # Backup replace if naming varies
    if aca_name == ac_name:
        aca_name = ac_name.replace('.pkl', '_ACA.pkl')
        
    output_path = os.path.join(DATA_FOLDER, aca_name)
    
    with open(output_path, 'wb') as f:
        pickle.dump(df_aca, f)
    
    print(f"   ✨ Created: {aca_name}")
    print(f"   📊 Final Shape: {df_aca.shape}\n")

print("✅ All files aligned and saved successfully.")
```

```python
import pandas as pd
import pickle
import os

DATA_FOLDER = 'cytotoxicity_data'
orig_name = 'tox21_all.pkl'
ac_name = 'tox21_all_AC.pkl'

def load_pkl(file_name):
    with open(os.path.join(DATA_FOLDER, file_name), 'rb') as f:
        return pickle.load(f, encoding='latin1')

# 1. LOAD
df_orig = load_pkl(orig_name)
df_ac = load_pkl(ac_name)

# 2. IDENTIFY SMILES COLUMN
s_col = next((c for c in df_orig.columns if 'smiles' in str(c).lower()), 'SMILES')

# 3. MANUAL ROW-BY-ROW CHECK AND SLICING
# We only take rows from df_ac that match the length of df_orig
# and verify alignment one last time
if len(df_orig) == len(df_ac):
    # Check if every single row matches at the exact same position
    matches = (df_orig[s_col].values == df_ac[s_col].values).all()
    
    if matches:
        print("✅ Alignment verified. Gluing columns now...")
        
        # Identify only the columns that are NOT in the original file
        new_cols = [c for c in df_ac.columns if c not in df_orig.columns]
        
        # Concatenate horizontally (axis=1) 
        # We reset the index to ensure they line up perfectly by row number
        df_aca = pd.concat([df_orig.reset_index(drop=True), 
                           df_ac[new_cols].reset_index(drop=True)], 
                          axis=1)
        
        print(f"✅ Success! New Row Count: {len(df_aca)} (Should be 2480)")
        
        # 4. SAVE
        aca_name = ac_name.replace('_AC.pkl', '_ACA.pkl')
        with open(os.path.join(DATA_FOLDER, aca_name), 'wb') as f:
            pickle.dump(df_aca, f)
            
    else:
        print("❌ SMILES do not match row-for-row. Manual glueing aborted.")
else:
    print(f"❌ Length mismatch: Orig({len(df_orig)}) vs AC({len(df_ac)})")
```

```python
import pandas as pd
import pickle
import os

# --- SETUP ---
DATA_FOLDER = 'cytotoxicity_data'
ACA_FILE = 'tox21_all_ACA.pkl'
FILE_PATH = os.path.join(DATA_FOLDER, ACA_FILE)

def analyze_ac_labels(file_path):
    if not os.path.exists(file_path):
        print(f"❌ File not found: {file_path}")
        return

    with open(file_path, 'rb') as f:
        df = pickle.load(f, encoding='latin1')

    # Ensure the column exists
    if 'ac_label' not in df.columns:
        print(f"❌ Column 'ac_label' not found in {ACA_FILE}")
        print(f"Available columns: {df.columns.tolist()}")
        return

    print(f"📊 Activity Cliff Statistics for {ACA_FILE}")
    print("=" * 50)

    # 1. Get raw counts of each unique value
    # value_counts() handles the frequency of each label
    label_stats = df['ac_label'].value_counts(dropna=False)

    # 2. Get percentages
    label_perc = df['ac_label'].value_counts(normalize=True, dropna=False) * 100

    # 3. Combine into a neat summary table
    summary = pd.DataFrame({
        'Count': label_stats,
        'Percentage (%)': label_perc.round(2)
    })

    print(summary)
    
    # 4. Specific Insights
    total_molecules = len(df)
    nan_count = df['ac_label'].isna().sum()
    
    print("-" * 50)
    print(f"Total Rows Analyzed: {total_molecules}")
    print(f"Molecules without labels (NaN): {nan_count}")

# Run the analysis
analyze_ac_labels(FILE_PATH)
```

```python
import os
import pandas as pd
import numpy as np
import deepchem as dc
from deepchem.molnet.load_function.molnet_loader import _MolnetLoader

class _ToxPKLLoader(_MolnetLoader):
    def create_dataset(self) -> dc.data.Dataset:
        pass

    def load_dataset(self, name: str, reload: bool = True):
        files = {'train': "HEK293_train_BM.pkl", 'val': "HEK293_test_BM.pkl", 'test': "tox21_all.pkl"}
        datasets = []

        for split, filename in files.items():
            path = os.path.join(self.data_dir, filename)
            if not os.path.exists(path):
                print(f"Skipping {split}: {path} not found.")
                continue
            
            df = pd.read_pickle(path)
            # Use ECFP featurizer passed from the call function
            features = self.featurizer.featurize(df['SMILES'].tolist())
            
            # Filter out failures (ECFP returns an empty array for failed SMILES)
            valid_inds = [i for i, f in enumerate(features) if f.size > 0]
            
            if len(valid_inds) == 0:
                print(f"Warning: No valid molecules in {split}")
                continue

            # ECFP features are typically (n_samples, n_bits)
            clean_features = np.stack([features[i] for i in valid_inds])
            clean_labels = df['Outcome'].values[valid_inds].reshape(-1, 1)
            clean_ids = np.array(df['SMILES'].tolist())[valid_inds]

            ds = dc.data.DiskDataset.from_numpy(X=clean_features, y=clean_labels, ids=clean_ids)
            datasets.append(ds)
            print(f"Loaded {split}: {len(ds)} samples.")

        return ['Outcome'], tuple(datasets), []

def load_custom_ecfp_tox(data_dir='cytotoxicity_data', size=1024):
    # Directly initialize the ECFP featurizer
    featurizer = dc.feat.CircularFingerprint(size=size, radius=2)
    
    loader = _ToxPKLLoader(
        featurizer=featurizer,
        splitter=None,
        transformer_generators=[],
        tasks=['Outcome'],
        data_dir=data_dir,
        save_dir='./save_ecfp'
    )
    return loader.load_dataset('custom_tox_ecfp', reload=False)
```

```python
import sys
import types
import numpy as np
import pandas as pd
import pickle
import os

# 1. ESTABLISH THE BRIDGE (Fixes NumPy 2.x compatibility)
_core = types.ModuleType("numpy._core")
_core.__path__ = [] 
sys.modules['numpy._core'] = _core
sys.modules['numpy._core.numeric'] = np.core.numeric
sys.modules['numpy._core.multiarray'] = np.core.multiarray
sys.modules['numpy._core.umath'] = np.core.umath

DATA_FOLDER = 'cytotoxicity_data'
files = ['HEK293_train_BM.pkl', 'HEK293_test_BM.pkl', 'tox21_all.pkl']
# The essential columns for the DeepChem pipeline
IMPORTANT_COLUMNS = ['SMILES', 'Outcome']

def pkl_to_csv_bridge(folder, file_name):
    file_path = os.path.join(folder, file_name)
    csv_path = file_path.replace(".pkl", ".csv")
    
    if not os.path.exists(file_path):
        print(f"Skipping: {file_name} not found.")
        return

    # 2. LOAD USING THE BRIDGE
    with open(file_path, 'rb') as f:
        # encoding='latin1' is critical for NumPy array reconstruction
        data = pickle.load(f, encoding='latin1')
    
    # 3. EXTRACT AND STORE
    if isinstance(data, pd.DataFrame):
        # Keep only SMILES and Outcome to ensure clean CSV storage
        df_to_save = data[IMPORTANT_COLUMNS]
        df_to_save.to_csv(csv_path, index=False)
        print(f"✅ Converted {file_name} -> {os.path.basename(csv_path)} ({len(df_to_save)} rows)")
    else:
        print(f"❌ Error: {file_name} did not load as a DataFrame.")

# Execute conversion
for f in files:
    pkl_to_csv_bridge(DATA_FOLDER, f)
```

```python
import pandas as pd
import numpy as np
import deepchem as dc
import os

def process_and_inspect_csv(file_path, featurizer_type='ECFP'):
    print(f"\n{'='*40}")
    print(f"Inspecting: {os.path.basename(file_path)}")
    print(f"{'='*40}")
    
    # 1. Load the "Clean" Data
    df = pd.read_csv(file_path)
    
    # 2. General Statistics (Native NumPy 2.0 types)
    total = len(df)
    actives = df['Outcome'].sum()
    inactives = total - actives
    ratio = inactives / actives if actives > 0 else 0
    
    print(f"Total Molecules: {total}")
    print(f"Class Distribution: {actives} Actives / {inactives} Inactives")
    print(f"Imbalance Ratio: 1 : {ratio:.2f}")

    # 3. Initialize Featurizer
    if featurizer_type == 'ECFP':
        featurizer = dc.feat.CircularFingerprint(size=1024, radius=2)
    else:
        featurizer = dc.feat.MolGraphConvFeaturizer()

    # 4. Featurize (This generates NumPy 2.0.0 native arrays)
    print(f"Featurizing with {featurizer_type}...")
    features = featurizer.featurize(df['SMILES'].tolist())
    
    # Filter out any failed SMILES
    valid_inds = [i for i, f in enumerate(features) if hasattr(f, 'size') and f.size > 0]
    
    X = np.array([features[i] for i in valid_inds])
    y = df['Outcome'].values[valid_inds].reshape(-1, 1)
    ids = df['SMILES'].values[valid_inds]

    # 5. Create DeepChem DiskDataset
    dataset = dc.data.DiskDataset.from_numpy(X=X, y=y, ids=ids)
    
    print(f"Final Dataset Shape: X={dataset.X.shape}, y={dataset.y.shape}")
    return dataset

# --- EXECUTION ---
DATA_FOLDER = 'cytotoxicity_data'
csv_files = ['HEK293_train_BM.csv', 'HEK293_test_BM.csv', 'tox21_all.csv']

processed_datasets = {}
for f in csv_files:
    full_path = os.path.join(DATA_FOLDER, f)
    if os.path.exists(full_path):
        split_name = f.split('.')[0]
        processed_datasets[split_name] = process_and_inspect_csv(full_path)
```

```python
import pandas as pd
import os

# Configuration
DATA_FOLDER = 'cytotoxicity_data'
FILES = ['HEK293_train_BM.pkl', 'HEK293_test_BM.pkl', 'tox21_all.pkl']
IMPORTANT_COLUMNS = ['SMILES', 'Outcome']

def convert_to_csv(folder, file_name):
    file_path = os.path.join(folder, file_name)
    csv_path = file_path.replace(".pkl", ".csv")
    
    if not os.path.exists(file_path):
        print(f"File not found: {file_path}")
        return

    # Direct load: This will raise an error if NumPy versions/paths mismatch
    df = pd.read_pickle(file_path)
    
    # Select essential columns
    df_clean = df[IMPORTANT_COLUMNS]
    
    # Save as CSV
    df_clean.to_csv(csv_path, index=False)
    print(f"Successfully converted: {file_name} -> {os.path.basename(csv_path)}")

# Run conversion
for f in FILES:
    convert_to_csv(DATA_FOLDER, f)
```
